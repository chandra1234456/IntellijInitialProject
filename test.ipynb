# Kotlin OOPS Interview Notes

---

# 1. WHAT IS OOPS?

**OOPS (Object-Oriented Programming System)** is a programming paradigm that organizes software around objects.

## Four Pillars

1. Encapsulation
2. Inheritance
3. Polymorphism
4. Abstraction

## Benefits

* Code Reusability
* Scalability
* Maintainability
* Security
* Better Code Organization

---

# 2. ENCAPSULATION

## Definition

Encapsulation is the process of binding data and methods together while hiding internal implementation details.

## Example

```kotlin
class BankAccount(private var balance: Double) {

    fun deposit(amount: Double) {
        balance += amount
    }

    fun getBalance() = balance
}
```

## Advantages

* Data Hiding
* Security
* Controlled Access
* Better Maintainability

### Interview Question

**Why make variables private?**

**Answer:**
To prevent direct modification and ensure access only through controlled methods.

### Real-Time Example

Repository pattern hides API and database implementation from ViewModel.

---

# 3. INHERITANCE

## Definition

Inheritance allows one class to acquire properties and behaviors of another class.

## Example

```kotlin
open class Animal {

    fun eat() {
        println("Eating")
    }
}

class Dog : Animal() {

    fun bark() {
        println("Barking")
    }
}
```

## Usage

```kotlin
val dog = Dog()

dog.eat()
dog.bark()
```

## Advantages

* Code Reusability
* Reduced Duplication
* Easy Maintenance

### Interview Question

**Why use open keyword?**

**Answer:**
Kotlin classes are final by default. Use `open` to allow inheritance.

# 4. POLYMORPHISM

## Definition

One interface, multiple implementations.

## A. Compile-Time Polymorphism (Method Overloading)

```kotlin
class Calculator {

    fun add(a: Int, b: Int) = a + b

    fun add(a: Double, b: Double) = a + b
}
```

## B. Runtime Polymorphism (Method Overriding)

```kotlin
open class Animal {

    open fun sound() {
        println("Animal Sound")
    }
}

class Dog : Animal() {

    override fun sound() {
        println("Bark")
    }
}
```

## Usage

```kotlin
val animal: Animal = Dog()

animal.sound()
```

### Output

```text
Bark
```

### Interview Question

**What enables Runtime Polymorphism?**

**Answer:**
Method Overriding + Dynamic Dispatch.

## Benefits

* Flexibility
* Loose Coupling
* Extensibility

---

# 5. ABSTRACTION

## Definition

Showing only essential information while hiding implementation details.

## A. Abstract Class

```kotlin
abstract class Vehicle {

    abstract fun start()

    fun stop() {
        println("Stopped")
    }
}

class Car : Vehicle() {

    override fun start() {
        println("Car Started")
    }
}
```

## B. Interface

```kotlin
interface Payment {

    fun pay(amount: Double)
}

class UpiPayment : Payment {

    override fun pay(amount: Double) {
        println("Paid using UPI")
    }
}
```

## Benefits

* Reduces Complexity
* Improves Security
* Flexible Design

---

# 6. ABSTRACT CLASS vs INTERFACE

| Abstract Class       | Interface                      |
| -------------------- | ------------------------------ |
| Can have state       | No constructor                 |
| Can have constructor | Multiple inheritance supported |
| Single inheritance   | Defines contract               |
| Common functionality | Behavior contract              |

### Interview Answer

Use **Abstract Class** when classes share common implementation.

Use **Interface** when classes need a common contract.

# 7. METHOD OVERLOADING vs METHOD OVERRIDING

## Overloading

```kotlin
fun add(a: Int, b: Int)
fun add(a: Double, b: Double)
```

### Features

* Same Method Name
* Different Parameters
* Compile-Time Polymorphism

## Overriding

```kotlin
open fun show()

override fun show()
```

### Features

* Same Method Signature
* Different Implementation
* Runtime Polymorphism

# 8. COMPOSITION vs INHERITANCE

## Composition

```kotlin
class Engine

class Car(
    private val engine: Engine
)
```

## Inheritance

```kotlin
open class Vehicle

class Car : Vehicle()
```

### Interview Answer

Composition is preferred because:

* Loosely Coupled
* Easy Testing
* Better Flexibility
* Better Maintainability

### Industry Rule

> Favor Composition over Inheritance

# 9. DATA CLASS

```kotlin
data class User(
    val id: Int,
    val name: String
)
```

## Automatically Generates

* equals()
* hashCode()
* toString()
* copy()
* componentN()

## Usage

```kotlin
val user1 = User(1, "John")

val user2 = user1.copy(
    name = "Mike"
)
```

### Interview Question

**Why Data Class?**

**Answer:**
Reduces boilerplate code.

# 11. COMPANION OBJECT

```kotlin
class User {

    companion object {

        const val MAX_AGE = 60

        fun create() = User()
    }
}
```

## Usage

```kotlin
User.MAX_AGE

User.create()
```

### Interview Answer

Companion Object provides class-level members and works similar to Java static.

# 12. SEALED CLASS

```kotlin
sealed class Result {

    data class Success(
        val data: String
    ) : Result()

    data class Error(
        val message: String
    ) : Result()

    object Loading : Result()
}
```

## Usage

```kotlin
when(result) {

    is Result.Success -> {}

    is Result.Error -> {}

    Result.Loading -> {}
}
```

## Benefits

* Type Safety
* Exhaustive when
* State Management

# 14. MULTIPLE INHERITANCE

## Class

Not Supported

## Interface

Supported

```kotlin
interface A {
    fun show()
}

interface B {
    fun show()
}

class C : A, B {

    override fun show() {
        println("Implemented")
    }
}
```

# 15. MOST ASKED INTERVIEW QUESTIONS

### Q1. What are the four pillars of OOPS?

Encapsulation, Inheritance, Polymorphism, Abstraction.

### Q2. Why are Kotlin classes final by default?

To prevent unintended inheritance and encourage composition.

### Q3. Difference between Abstract Class and Interface?

Abstract Class can have state and constructor.

Interface defines behavior contract.

### Q4. Difference between Overloading and Overriding?

Overloading → Compile Time

Overriding → Runtime

### Q5. What is Runtime Polymorphism?

Method call resolved at runtime using overriding.

### Q6. What is Encapsulation?

Hiding data and exposing controlled access methods.

### Q7. Why use Composition over Inheritance?

Less coupling, more flexibility, easier maintenance.

### Q8. What is a Data Class?

A class designed to hold data with autogenerated methods.

### Q9. What is a Companion Object?

Kotlin alternative for static members.

### Q10. What is a Sealed Class?

Restricted hierarchy used for type-safe state management.

# Constructor Interview Note

A constructor is a special method that runs automatically when an object is created.

### Key Points

* Constructor name must be the same as the class name.
* Constructors have no return type.
* Called automatically when an object is created.
* A class can have multiple constructors.
